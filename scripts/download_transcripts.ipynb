```python
%%bash
while read url; do
  echo "Processing $url";
  # if url ends with extension .m3u8 use ffmpeg, else youtube-dl else use curl to download video
  if [[ $url == *.m3u8 ]]; then
  #   filename=`basename "$url"`
  #   ffmpeg -i $url -c copy ${filename%.m3u8}.mp4
  # elif [[ $url == *"$youtube"* ]]; then
    youtube-dl --format 5605 $url;
  elif [[ -n "${url}" ]]; then
    curl -O $url;
  fi
done < urls.txt
```

Compress MP4 files

```python
%%bash

mkdir -p videos

for i in *.mp4;
do
    echo "Processing $i -> ${i%%.mp4}"
    ffmpeg -i "$i" -vcodec libx265 -crf 28 "videos/${i%%.mp4}_h265.mp4"
done
```

```python
!ls -alt videos/*.mp4
```

```python
!tar czf videos.tar.gz videos/*_h265.mp4
```

Generate transcripts

```python
!pip install -q -U openai-whisper
```

```python
%%bash

mkdir -p audios

for i in *.mp4;
do
    echo "Processing $i -> ${i%%.mp4}"
    ffmpeg -i "$i" "audios/${i%%.mp4}.mp3"
done
```

```python
%%bash

mkdir -p transcripts

for i in audios/*.mp3;
do
    echo "Processing $i -> ${i%%.txt}"
    whisper "$i" --model base --language en --output_format txt --output_dir transcripts
done
```

```python
!tar czf transcripts.tar.gz transcripts/
```
